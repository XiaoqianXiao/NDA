```python
#helper functions
# Calculate interview_age in months from date_of_birth to interview_date
# Age is rounded to chronological month: 15 days = 0 months, 16 days = 1 month
from datetime import datetime

# Calculate interview_age in months
def calculate_age_months(dob_str, interview_date_str):
    if pd.isna(dob_str) or pd.isna(interview_date_str):
        return None
    try:
        # Parse dates (date_of_birth is in yyyy/mm/dd format)
        dob = pd.to_datetime(dob_str, format='%Y/%m/%d')
        interview_date = pd.to_datetime(interview_date_str)
        
        # Calculate total days difference
        delta = interview_date - dob
        total_days = delta.days
        
        # Calculate base months (years * 12 + month difference)
        years = interview_date.year - dob.year
        months = interview_date.month - dob.month
        base_months = years * 12 + months
        
        # Adjust based on day of month
        # If interview day is earlier than birth day, we're in the previous month
        if interview_date.day < dob.day:
            base_months -= 1
            # Calculate days into the current month
            days_into_month = interview_date.day
        else:
            # Calculate days into the current month from birth day
            days_into_month = interview_date.day - dob.day
        
        # Apply rounding rule: 15 days = 0 months, 16 days = 1 month
        # If days_into_month >= 16, add 1 month
        if days_into_month >= 16:
            base_months += 1
        
        return max(0, base_months)  # Ensure non-negative
    except:
        return None

# Dictionary for phenotype_description mapping
dic_phenotype_description = {
    'HC': 'Matched healthy controls',
    'SAD': 'Patients with Anxiety Disorders',
    'BDD': 'Patients with Body Dysmorphic Disorder disorders',
    'SAD/BDD': 'Patients with Anxiety Disorders and Body Dysmorphic Disorder disorders'
}

# Dictionary for Race 1=American Indian/Alaska Native; 2=Asian; 3=Black or African American; 4=awaiian or Pacific Islander; 5=White; 6=More than one race; 7=Unknown or not reported
dic_race = {
    1: "American Indian/Alaska Native",
    2: "Asian",
    3: "Black or African American",
    4: "Hawaiian or Pacific Islander",
    5: "White",
    6: "More than one race",
    7: "Unknown or not reported"
}

dic_mdd_lp = {
    1:0,
    3:1
}
```

```python
# #10. image03
questionaire_name = 'image03'
results_file = os.path.join(result_dir, f'{questionaire_name}.csv')
# --- File Loading ---
template_path = [f for f in glob.glob(os.path.join(approved_dir, '*.csv')) if questionaire_name.lower() in os.path.basename(f).lower()][0]
defination_path = [f for f in glob.glob(os.path.join(defination_dir, '*.csv')) if questionaire_name.lower() in os.path.basename(f).lower()][0]
df_def = pd.read_csv(defination_path)
dtype_map = dict(zip(df_def['ElementName'], df_def['DataType']))
redcap_file_path = [f for f in glob.glob(os.path.join(redcap_file_dir, '*.csv')) if questionaire_name.lower() in os.path.basename(f).lower()][0]
df_redcap_ori = pd.read_csv(redcap_file_path)
target_column = "mri_exit_date"
df_redcap = df_redcap_ori[df_redcap_ori[target_column].notna()]

extract_file = os.path.join(sumission_dir, 'NDA_image03_extracted.csv')
df_extract_ori = pd.read_csv(extract_file)
# ---------------------------------------------------------
# clean MRI data 1: NDA_image03_extracted.csv
# ---------------------------------------------------------
# sub-304_ses-baseline1 is sub-304_ses-repeatbaseline
# sub-132_ses-baseline is not usable, get rid of them; set sub-132_ses-baseline1 to sub-132_ses-baseline
# sub-141_ses-baseline is not usable, get rid of them; set sub-141_ses-baseline1 to sub-141_ses-baseline
# sub-326_ses-T12 is sub-325_ses-T12
# sub-326_ses-T121 is sub-326_ses-T12
# --- Configuration & Variables ---
# Define the column names as per your dataset
sub_col = "ID"
ses_col = "visit"
# Dropping rules: (ID, visit) pairs to remove
records_to_drop = [
    ("132", "baseline"),
    ("141", "baseline")
]
# Mapping rules for corrections: (current_ID, current_visit) -> (new_ID, new_visit)
mapping_rules = {
    ("304", "baseline1"): ("304", "repeatbaseline"),
    ("132", "baseline1"): ("132", "baseline"),
    ("141", "baseline1"): ("141", "baseline"),
    ("326", "T12"):       ("325", "T12"),
    ("326", "T121"):      ("326", "T12")
}
# --- Data Processing ---
# 1. Standardize column types to string to ensure exact matching
df_extract_ori[sub_col] = df_extract_ori[sub_col].astype(str)
df_extract_ori[ses_col] = df_extract_ori[ses_col].astype(str)
# 2. Drop the unusable sessions first
for sub_id, ses_id in records_to_drop:
    df_extract_ori = df_extract_ori[~((df_extract_ori[sub_col] == sub_id) & (df_extract_ori[ses_col] == ses_id))]
# 3. Apply the corrections mapping
def apply_corrections(row):
    current_key = (row[sub_col], row[ses_col])
    if current_key in mapping_rules:
        new_sub, new_ses = mapping_rules[current_key]
        row[sub_col] = new_sub
        row[ses_col] = new_ses
    return row
df_extract = df_extract_ori.apply(apply_corrections, axis=1).copy()

# ---------------------------------------------------------
#  [AUDIT 1]Check: Are we missing IDs right from the start?
# ---------------------------------------------------------
df_redcap['record_id'] = df_redcap['record_id'].astype(str).str.strip()
df_extract['src_subject_id'] = df_extract['src_subject_id'].astype(str).str.strip()
redcap_ids = set(df_redcap['record_id'].unique())
extract_ids = set(df_extract['src_subject_id'].unique())
missing_in_extract = redcap_ids - extract_ids
if missing_in_extract:
    print(f"\n[AUDIT 1] {len(missing_in_extract)} IDs exist in REDCap but are NOT in your extracted file.")
    print(f"Sample missing: {list(missing_in_extract)[:5]}")
else:
    print("\n[AUDIT 1] All REDCap IDs are present in the extraction file.")
# ---------------------------------------------------------
df_template = pd.read_csv(template_path, header=1)
first_row_template = pd.read_csv(template_path, header=None, nrows=1, names=df_template.columns)
# Initial DataFrame Setup
df = df_template.copy() 
common_cols = df_template.columns.intersection(df_extract.columns)
# Preserve original index to look up 'ID' later
df = df_extract[common_cols].copy()
# Merge GUID from df_meta
guid_map = df_meta.set_index('ID')['GUID'].to_dict()
df['src_subject_id'] = df['src_subject_id'].astype(str).str.strip()
df['subjectkey'] = df['src_subject_id'].map(guid_map)
# ---------------------------------------------------------
# [AUDIT 2]Check for missing GUIDs (Report Unique SubID)
# ---------------------------------------------------------
missing_guids = df[df['subjectkey'].isna()]
if not missing_guids.empty:
    # Use indices to lookup the original 'ID' in df_extract
    affected_indices = missing_guids.index
    unique_missing_ids = df_extract.loc[affected_indices, 'ID'].unique() if 'ID' in df_extract.columns else ["ID_COL_MISSING"]
    
    print(f"\n[AUDIT 2] {len(missing_guids)} rows failed to map a GUID.")
    print(f"   -> Affects {len(unique_missing_ids)} Unique Subject IDs: {unique_missing_ids}")
# ---------------------------------------------------------
# 1. Ensure IDs are strings
df['src_subject_id'] = df['src_subject_id'].astype(str).str.strip()
df_redcap['record_id'] = df_redcap['record_id'].astype(str).str.strip()
# 2. Optimization: Group REDCap data by ID
redcap_grouped = df_redcap.groupby('record_id')

# Precise mapping relationship variable between df_extract['visit'] and df_redcap['redcap_event_name']
visit_to_event_map = {
    'baseline': ['baseline_1_arm_1', 'baseline_1_arm_2'],
    'repeatbaseline': ['repeat_baseline_arm_1', 'repeat_baseline_arm_2'],
    't6': ['treatment_session_arm_1f'],
    't12': ['treatment_session_arm_1l', 't12_arm_2'],
}

def get_mri_date(row):
    subject_id = row['src_subject_id']
    visit_val = row['visit']
    
    if pd.isna(visit_val):
        return row['interview_date']
        
    visit_clean = str(visit_val).strip().lower()
    
    # Check if the extracted visit value exists in our explicit mapping rule
    if visit_clean in visit_to_event_map:
        target_events = visit_to_event_map[visit_clean]
        
        if subject_id in redcap_grouped.groups:
            subj_data = redcap_grouped.get_group(subject_id)
            # Find rows matching the specified target REDCap events
            match = subj_data[
                subj_data['redcap_event_name']
                .fillna('')
                .str.strip()
                .str.lower()
                .isin(target_events)
            ]
            if not match.empty:
                return match.iloc[0]['mri_exit_date']
                
    return row['interview_date']

# 3. Apply logic
df['interview_date'] = df.apply(get_mri_date, axis=1)
# Merge SEX & DOB
sex_map = df_meta.set_index('ID')['SEX'].to_dict()
df['sex'] = df['src_subject_id'].map(sex_map)
dob_map = df_meta.set_index('ID')['date_of_birth'].to_dict()
df['date_of_birth'] = df['src_subject_id'].map(dob_map)
df['interview_age'] = df.apply(
    lambda row: calculate_age_months(row['date_of_birth'], row['interview_date']), 
    axis=1
)
df['interview_age'] = df['interview_age'].astype('Int64')
df['experiment_id'] = df['experiment_id'].astype('Int64')
df['image_extent4'] = df['image_extent4'].astype('Int64')
df = df.astype(object)
df = df.where(pd.notnull(df), None)
# ---------------------------------------------------------
# [AUDIT 3]'MR structural (T2)' (Report Unique SubID)
# ---------------------------------------------------------
# Identify rows to drop
drop_mask_t2 = df['scan_type'].astype(str).str.contains('MR structural (T2)', case=False, na=False, regex=False)
rows_t2 = df[drop_mask_t2]
if not rows_t2.empty:
    affected_indices = rows_t2.index
    unique_t2_ids = df_extract.loc[affected_indices, 'ID'].unique() if 'ID' in df_extract.columns else ["ID_COL_MISSING"]
    
    print(f"\n[AUDIT 3]  Filtering out {len(rows_t2)} rows containing 'MR structural (T2)'.")
    print(f"   -> Affects {len(unique_t2_ids)} Unique Subject IDs: {unique_t2_ids}")
# Actual Drop
df = df[~drop_mask_t2].copy()
# ---------------------------------------------------------
# [AUDIT 4]'localizer scan' (Report Unique SubID)
# ---------------------------------------------------------
# Identify rows to drop
drop_mask_localizer = df['scan_type'].astype(str).str.contains('localizer scan', case=False, na=False)
rows_localizer = df[drop_mask_localizer]
if not rows_localizer.empty:
    affected_indices = rows_localizer.index
    unique_loc_ids = df_extract.loc[affected_indices, 'ID'].unique() if 'ID' in df_extract.columns else ["ID_COL_MISSING"]
    
    print(f"\n[AUDIT 4] Filtering out {len(rows_localizer)} rows containing 'localizer scan'.")
    print(f"   -> Affects {len(unique_loc_ids)} Unique Subject IDs: {unique_loc_ids}")
# Actual Drop (Keep the inverse of the drop mask)
df = df[~drop_mask_localizer].copy()
# ---------------------------------------------------------
# [AUDIT 5]Missing Age/Date (Report Unique SubID)
# ---------------------------------------------------------
# Filter rows that satisfy the drop condition
rows_to_drop = df[df['interview_age'].isnull() | df['interview_date'].isnull()]
if not rows_to_drop.empty:
    affected_indices = rows_to_drop.index
    unique_dropped_ids = df_extract.loc[affected_indices, 'ID'].unique() if 'ID' in df_extract.columns else ["ID_COL_MISSING"]
    print(f"\n[AUDIT 5]  Dropping {len(rows_to_drop)} rows due to missing Age or Date.")
    print(f"   -> Affects {len(unique_dropped_ids)} Unique Subject IDs: {unique_dropped_ids}")
else:
    print("\n[AUDIT 5] No rows dropped due to missing Age/Date.")
# Actual Drop
df = df.dropna(subset=['interview_age', 'interview_date'])
# ---------------------------------------------------------
# Final Formatting
# ---------------------------------------------------------
df['interview_date'] = pd.to_datetime(df['interview_date'])
df['interview_date'] = df['interview_date'].dt.strftime('%m/%d/%Y')

df = df[df_template.columns.tolist()]
df = pd.concat([first_row_template, df], ignore_index=True)
df.iloc[0] = df.iloc[0].fillna('')

df.iloc[:1].to_csv(results_file, index=False, header=False, mode='w')
df.iloc[1:].to_csv(results_file, index=False, header=True, mode='a')

print(f"\n[DONE] Final file saved to: {results_file}")
```

```python
# keep only the defaced T1w images
# Configuration
root_dir = os.path.join(sumission_dir, 'final/image')
target_subfolder = 'anat'
target_tag = 'desc-defaced_T1w'

# SET TO FALSE TO ACTUALLY DELETE
#dry_run = True
dry_run = False

print(f"--- STARTING CLEANUP (Dry Run: {dry_run}) ---")
print(f"Scanning: {root_dir}\n")

cleaned_folders = 0
deleted_files = 0

for dirpath, dirnames, filenames in os.walk(root_dir):
    # Only process directories named 'anat'
    if os.path.basename(dirpath) == target_subfolder:
        cleaned_folders += 1
        print(f"Checking folder: {dirpath}")
        
        # 1. Identify the "Master" files (the defaced NIfTIs)
        # We look for files containing 'desc-defaced_T1w' ending in .nii or .nii.gz
        defaced_niftis = [f for f in filenames if target_tag in f and (f.endswith('.nii') or f.endswith('.nii.gz'))]
        
        # 2. Build the Whitelist (Files we MUST keep)
        whitelist = set()
        
        for nifti in defaced_niftis:
            # Add the image itself to whitelist
            whitelist.add(nifti)
            
            # Derive the corresponding JSON name
            # Logic: Remove 'desc-defaced_' and switch extension to .json
            # e.g., sub-102_..._desc-defaced_T1w.nii.gz -> sub-102_..._T1w.json
            
            # Remove extension first
            if nifti.endswith('.nii.gz'):
                stem = nifti[:-7]
            else:
                stem = nifti[:-4]
            
            # Remove the 'desc-defaced_' tag to match the original JSON
            # We use replace() to handle cases where it might be in the middle
            json_stem = stem.replace('desc-defaced_', '') 
            json_filename = json_stem + '.json'
            
            # If this JSON actually exists, add it to whitelist
            if json_filename in filenames:
                whitelist.add(json_filename)
        
        # 3. Delete everything NOT in the whitelist
        for file in filenames:
            if file not in whitelist:
                full_path = os.path.join(dirpath, file)
                
                if dry_run:
                    print(f"  [WOULD DELETE]: {file}")
                else:
                    os.remove(full_path)
                    print(f"  [DELETED]: {file}")
                deleted_files += 1
            else:
                # Optional: print what we are keeping just to be sure
                # print(f"  [KEEPING]: {file}")
                pass

print(f"\n--- SUMMARY ---")
print(f"Anat folders processed: {cleaned_folders}")
print(f"Files {'marked for deletion' if dry_run else 'deleted'}: {deleted_files}")

if dry_run and deleted_files > 0:
    print(">>> Logic look good? Set 'dry_run = False' and run again.")
```
